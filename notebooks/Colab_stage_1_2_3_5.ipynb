### Sanity Check - Rejected by QC Review

```python
import json
import os
from typing import Dict, Iterable, Iterator, List
def read_jsonl(path: str) -> List[Dict]:
    """Read a JSONL file into a list of dicts. Missing file -> empty list."""
    if not os.path.exists(path):
        return []
    rows: List[Dict] = []
    with open(path, encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if line:
                rows.append(json.loads(line))
    return rows
```

```python
# sanity-check thresholds: listen to a couple of QC failures
from IPython.display import Audio, display
rows = read_jsonl(f'{OUT}/dataset_manifest.jsonl')
print('pass/total :', sum(r['qc_passed'] for r in rows), '/', len(rows))
print('pass :', sum(r['qc_passed'] for r in rows))
print('fail:', sum(not r['qc_passed'] for r in rows))
```

```python
# Review Failed!!
fails = [r for r in rows if not r['qc_passed']][:] # Edit show much fail you want to Verify not exceeding total
for r in fails:
    print(f"ID:{r['utt_id']} | CER= {r.get('cer')} | speaker similarity ={r.get('speaker_sim')} | QC Reasons: {r['qc_reasons']}")
    display(Audio(f"{OUT}/{r['audio_filepath']}"))
```

```python
passes = [r for r in rows if r['qc_passed']][:5] # Edit show much fail you want to Verify not exceeding total
for r in passes:
    print(f"ID:{r['utt_id']} | CER= {r.get('cer')} | speaker similarity ={r.get('speaker_sim')} | QC Reasons: Passed")
    display(Audio(f"{OUT}/{r['audio_filepath']}"))
```
